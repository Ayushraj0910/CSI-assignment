# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic with conditional routing
- Tool integration
- Structured JSON output
- Comprehensive error handling


````python
import json

# 🛠️ TOOL 1: Calculator
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        # Remove 'calculate' keyword and clean the expression
        cleaned = expression.replace("calculate", "").strip()
        result = eval(cleaned)
        return str(result)
    except Exception as e:
        return f"Error in calculation: {str(e)}"
````

````python
# 🛠️ TOOL 2: Keyword Extractor
def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        # Remove 'keywords' keyword and 'from' if present
        cleaned_text = text.lower().replace("keywords", "").replace("from", "")
        words = cleaned_text.split()
        # Filter words longer than 4 characters and remove duplicates
        keywords = list(set([w.strip() for w in words if len(w.strip()) > 4]))
        return keywords[:5]  # Return top 5
    except Exception:
        return []
````

## 🤖 Agent Implementation - Complete Solution

### Routing Logic:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response
- All errors → error type response

````python
# 🤖 AGENT FUNCTION - COMPLETE IMPLEMENTATION
def agent(query: str) -> dict:
    """
    Smart Assistant Agent with conditional routing
    
    Args:
        query (str): User query string
        
    Returns:
        dict: Structured JSON response with 'type' and 'result' keys
    """
    
    # Convert to lowercase for pattern matching
    query_lower = query.lower()
    
    try:
        # Route 1: Math Calculation
        if "calculate" in query_lower:
            # Extract the mathematical expression
            expression = query_lower.replace("calculate", "").strip()
            
            # Validate that there's something to calculate
            if not expression or expression == "":
                return {
                    "type": "error",
                    "result": "No expression provided for calculation"
                }
            
            # Call calculator tool
            result = calculator(expression)
            
            # Check if calculation was successful
            if "Error" in result:
                return {
                    "type": "error",
                    "result": result
                }
            
            return {
                "type": "calculation",
                "result": result
            }
        
        # Route 2: Keyword Extraction
        elif "keywords" in query_lower:
            # Extract the text to process
            text_to_process = query_lower.replace("keywords", "").replace("extract", "").replace("from", "").strip()
            
            # Validate that there's text to extract keywords from
            if not text_to_process or text_to_process == "":
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction"
                }
            
            # Call keyword extractor tool
            keywords = extract_keywords(text_to_process)
            
            return {
                "type": "keywords",
                "result": keywords
            }
        
        # Route 3: General Query (Fallback)
        else:
            return {
                "type": "general",
                "result": f"Query received: '{query}'. I can help with calculations or keyword extraction."
            }
    
    except Exception as e:
        # Global error handler
        return {
            "type": "error",
            "result": f"An unexpected error occurred: {str(e)}"
        }
````

## 📦 Expected Output Format

```json
{
  "type": "calculation / keywords / general / error",
  "result": "<result_value>"
}
```

### Validation Checklist:
✅ Response is a valid dictionary  
✅ Contains 'type' and 'result' keys  
✅ Type is one of: calculation, keywords, general, error  
✅ Result is JSON serializable  
✅ No information loss in processing

````python
# 🧪 TEST CASES - VALIDATION SUITE

print("\n" + "="*70)
print("🧪 RUNNING TEST SUITE - VALIDATION CHECK")
print("="*70 + "\n")

test_queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

all_passed = True

for i, q in enumerate(test_queries, 1):
    print(f"Test {i}: {q}")
    print("-" * 70)
    
    response = agent(q)
    
    # Validation Checks
    validation_passed = True
    
    # Check 1: Response is a dictionary
    if not isinstance(response, dict):
        print(f"❌ FAILED: Response is not a dictionary")
        validation_passed = False
        all_passed = False
    else:
        print(f"✅ Response is a valid dictionary")
    
    # Check 2: Required keys present
    if "type" not in response or "result" not in response:
        print(f"❌ FAILED: Missing required keys. Got keys: {list(response.keys())}")
        validation_passed = False
        all_passed = False
    else:
        print(f"✅ Contains required keys: 'type' and 'result'")
    
    # Check 3: Type is valid
    valid_types = ["calculation", "keywords", "general", "error"]
    if response.get("type") not in valid_types:
        print(f"❌ FAILED: Invalid type '{response.get('type')}'. Must be one of: {valid_types}")
        validation_passed = False
        all_passed = False
    else:
        print(f"✅ Valid type: '{response.get('type')}'")
    
    # Check 4: JSON serializable
    try:
        json_str = json.dumps(response)
        print(f"✅ Response is valid JSON")
    except:
        print(f"❌ FAILED: Response is not JSON serializable")
        validation_passed = False
        all_passed = False
    
    # Print the actual response
    print(f"\nResponse:")
    print(json.dumps(response, indent=2))
    print("\n" + "="*70 + "\n")

# Final summary
if all_passed:
    print("🎉 ALL TESTS PASSED - Agent working correctly!")
else:
    print("⚠️ Some tests failed - review the agent implementation")
````

````python
# 🛡️ BONUS: ERROR HANDLING TESTS

print("\n" + "="*70)
print("🛡️ ERROR HANDLING VALIDATION")
print("="*70 + "\n")

error_test_cases = [
    ("Calculate abc + xyz", "error", "Invalid math expression"),
    ("Extract keywords from", "error", "Missing text for keyword extraction"),
    ("Calculate", "error", "Missing expression"),
    ("", "error", "Empty query"),
    ("Tell me about Python", "general", "General fallback"),
]

for query, expected_type, description in error_test_cases:
    response = agent(query)
    
    # Verify it returns error or general type
    is_valid = (response.get("type") == expected_type or 
                response.get("type") in ["error", "general"])
    
    status = "✅" if is_valid else "❌"
    print(f"{status} {description}")
    print(f"   Query: '{query}'")
    print(f"   Type: {response.get('type')}")
    print(f"   Result: {response.get('result')}\n")
````

````python
# 🎯 Interactive Mode

print("\n" + "="*70)
print("🎯 INTERACTIVE MODE - Enter queries (type 'exit' to stop)")
print("="*70)
print("\nTry queries like:")
print("  • 'Calculate 15 * 3'")
print("  • 'Extract keywords from Python is a great language'")
print("  • 'Tell me a joke'")
print("="*70 + "\n")

while True:
    try:
        user_input = input("Enter query (type 'exit' to stop): ").strip()
        
        if user_input.lower() == "exit":
            print("Goodbye!")
            break
        
        if not user_input:
            print("Please enter a valid query.\n")
            continue
        
        response = agent(user_input)
        print(f"\nResponse:")
        print(json.dumps(response, indent=2))
        print()
    
    except KeyboardInterrupt:
        print("\n\nInterrupted by user. Goodbye!")
        break
    except Exception as e:
        print(f"Error: {e}\n")
````
